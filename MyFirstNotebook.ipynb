```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.linspace(0, 2 * np.pi, 100) # Generate 100 points from 0 to 2*pi
y = np.sin(x) # Calculate sin(x)
```

```python
plt.figure(figsize=(8, 6))
plt.plot(x, y, label='sin(x)')
plt.title('Plot of sin(x)')
plt.xlabel('x')
plt.ylabel('sin(x)')
plt.grid(True)
plt.legend()
plt.show()
```
